```python
d4["Multiply"]= d4["s"]*d4["d"]
d4["Multiply>20"]=d4["Multiply"]>20
d4
```

```python
d4["Sum"]=d4["a"]+d4["s"]+d4["d"]
d4
d4["Diff(a-b)"]=d4["a"]-d4["s"]
d4
d4["Mul(a*b)"]=d4["a"]*d4["s"]
d4
d4["Div(a/b)"]=d4["a"]/d4["s"]
d4
```
